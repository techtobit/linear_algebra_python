```python
import numpy as np
A = np.array([
    [4, 2, 2],
    [2, 4, 2],
    [2, 2, 4]
])

A_Det = np.linalg.det(A)
print(round(A_Det))
```

```python
inv = np.linalg.inv(A)
print(inv)
```

```python
values, vectors = np.linalg.eig(A)
print(vectors)
```

```python
A = np.array([[2, 1],
              [1, 2]])

np.linalg.det(A)      # determinant
np.linalg.inv(A)      # inverse
np.linalg.eig(A)      # eigen
```
