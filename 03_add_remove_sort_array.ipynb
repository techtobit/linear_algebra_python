```python
import numpy as np #short array
a = np.array([2,8,5,8,3,4,1,0,9,6])
print(np.sort(a))
print(a.max())
print(a.min())
# print(a.sum())
```

```python
array_a = np.array([1,2,3,4,])
array_b = np.array([5,6,6,7,])

print(np.concatenate((array_a, array_b))) #  #adding two array
print(array_a + array_b) #additions
# print(array_a - array_b) #substraction
print( array_a * array_b) #multification
print(array_b / array_a) #divition
```
