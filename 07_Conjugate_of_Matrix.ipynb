```python
import numpy as np
A = np.array([
    [1+2j, 3-1j],
    [-2j, 4+5j]
])

conjugate = np.conj(A)
print(conjugate)
```

```python
#properties
# -----  --   --
# A + B = A  + B

A = np.array([
    [1+2j, 3-1j],
    [-2j, 4+5j] 
])

B = np.array([
    [2-1j, 1j],
    [1+3j, -2]
])

print(np.conj(A+B))
print("----------------")
A_puls_B = np.conj(A) + np.conj(B)
print(A_puls_B)
```
